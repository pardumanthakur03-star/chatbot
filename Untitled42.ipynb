```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import re
import nltk
from textblob import TextBlob
from wordcloud import WordCloud
from collections import Counter

nltk.download('stopwords')

from nltk.corpus import stopwords

df = pd.read_csv("/content/iphone.csv")

print("Dataset Shape:", df.shape)

print(df.head())
df = df[['reviewTitle',
         'reviewDescription',
         'ratingScore',
         'date',
         'country',
         'isVerified']]

df.dropna(inplace=True)

df['full_review'] = (
    df['reviewTitle'].astype(str) + " " +
    df['reviewDescription'].astype(str)
)

stop_words = set(stopwords.words('english'))

def clean_text(text):

    text = text.lower()

    # remove urls
    text = re.sub(r"http\S+", "", text)

    # remove numbers
    text = re.sub(r"\d+", "", text)

    # remove special characters
    text = re.sub(r"[^a-zA-Z ]", "", text)

    # remove extra spaces
    text = re.sub(r"\s+", " ", text)

    # remove stopwords
    words = text.split()

    words = [word for word in words if word not in stop_words]

    return " ".join(words)

df['clean_review'] = df['full_review'].apply(clean_text)

print(df[['full_review', 'clean_review']].head())

# =========================
# 7. SENTIMENT ANALYSIS
# =========================

def get_sentiment(text):

    score = TextBlob(text).sentiment.polarity

    if score > 0:
        return "Positive"

    elif score < 0:
        return "Negative"

    else:
        return "Neutral"

df['Sentiment'] = df['clean_review'].apply(get_sentiment)

# =========================
# 8. SENTIMENT SCORE
# =========================

def polarity_score(text):
    return TextBlob(text).sentiment.polarity

df['Sentiment_Score'] = df['clean_review'].apply(polarity_score)

# =========================
# 9. DISPLAY SENTIMENT COUNTS
# =========================

print("\nSentiment Distribution:\n")

print(df['Sentiment'].value_counts())

# =========================
# 10. BAR CHART
# =========================

plt.figure(figsize=(8,5))

df['Sentiment'].value_counts().plot(kind='bar')

plt.title("Sentiment Distribution")

plt.xlabel("Sentiment")

plt.ylabel("Count")

plt.show()

# =========================
# 11. PIE CHART
# =========================

plt.figure(figsize=(7,7))

df['Sentiment'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%'
)

plt.title("Sentiment Percentage")

plt.ylabel("")

plt.show()

# =========================
# 12. WORD CLOUD
# =========================

text = " ".join(df['clean_review'])

wordcloud = WordCloud(
    width=1000,
    height=500,
    background_color='white'
).generate(text)

plt.figure(figsize=(12,6))

plt.imshow(wordcloud)

plt.axis("off")

plt.title("Most Common Words")

plt.show()

# =========================
# 13. TOP CUSTOMER ISSUES
# =========================

negative_reviews = df[df['Sentiment'] == 'Negative']

negative_words = " ".join(
    negative_reviews['clean_review']
).split()

common_issues = Counter(negative_words).most_common(10)

print("\nTop Customer Complaints:\n")

for word, count in common_issues:
    print(word, ":", count)

# =========================
# 14. KPI METRICS
# =========================

total_reviews = len(df)

positive_reviews = len(
    df[df['Sentiment'] == 'Positive']
)

negative_reviews_count = len(
    df[df['Sentiment'] == 'Negative']
)

neutral_reviews = len(
    df[df['Sentiment'] == 'Neutral']
)

positive_percentage = (
    positive_reviews / total_reviews
) * 100

negative_percentage = (
    negative_reviews_count / total_reviews
) * 100

print("\n========== KPI METRICS ==========")

print("Total Reviews:", total_reviews)

print("Positive Reviews:", positive_reviews)

print("Negative Reviews:", negative_reviews_count)

print("Neutral Reviews:", neutral_reviews)

print("Positive %:", round(positive_percentage,2))

print("Negative %:", round(negative_percentage,2))

# =========================
# 15. TREND ANALYSIS
# =========================

# convert date column
df['date'] = pd.to_datetime(
    df['date'],
    errors='coerce'
)

monthly_trend = df.groupby(
    df['date'].dt.to_period('M')
)['Sentiment_Score'].mean()

monthly_trend.plot(figsize=(12,5))

plt.title("Monthly Sentiment Trend")

plt.xlabel("Month")

plt.ylabel("Average Sentiment Score")

plt.show()

# =========================
# 16. BUSINESS INSIGHTS
# =========================

print("\n========== BUSINESS INSIGHTS ==========\n")

if positive_percentage > 70:
    print("Customers are highly satisfied.")

if negative_percentage > 20:
    print("Negative reviews are increasing.")

print("\nTop recommendations:")

print("1. Improve charging accessories.")

print("2. Improve camera usability.")

print("3. Improve software experience.")

print("4. Enhance customer support.")

print("5. Improve pricing strategy.")

# =========================
# 17. EXPORT RESULTS
# =========================

df.to_csv(
    "sentiment_output.csv",
    index=False
)

print("\nSentiment file exported successfully.")

# =========================
# 18. AI BUSINESS CHATBOT
# =========================

def chatbot(query):

    query = query.lower()

    # overall sentiment
    if "overall sentiment" in query:

        return str(
            df['Sentiment'].value_counts()
        )

    # positive reviews
    elif "positive reviews" in query:

        return (
            "Positive Reviews: " +
            str(positive_reviews)
        )

    # negative reviews
    elif "negative reviews" in query:

        return (
            "Negative Reviews: " +
            str(negative_reviews_count)
        )

    # top complaints
    elif "top complaints" in query:

        return (
            "Main complaints are: "
            "charger issue, camera issue, "
            "software issue."
        )

    # improvement suggestions
    elif "improvement" in query:

        return (
            "Suggested improvements: "
            "better charger support, "
            "camera update, software optimization."
        )

    # trend
    elif "trend" in query:

        return (
            "Customer sentiment is generally positive."
        )

    # help
    elif "help" in query:

        return (
            "Ask questions like:\n"
            "- overall sentiment\n"
            "- top complaints\n"
            "- positive reviews\n"
            "- improvement suggestions\n"
            "- trend"
        )

    else:
        return "Sorry, I don't understand."

# =========================
# 19. RUN CHATBOT
# =========================

print("\n========== AI BUSINESS CHATBOT ==========\n")

print("Type 'exit' to stop chatbot.")

while True:

    user_input = input("\nAsk Question: ")

    if user_input.lower() == "exit":
        print("Chatbot Ended.")
        break

    response = chatbot(user_input)

    print("\nBot:", response)


```
